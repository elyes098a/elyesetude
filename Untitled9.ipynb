```python
# -*- coding: utf-8 -*-
import os
import logging
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

logging.basicConfig(level=logging.INFO, format='%(asctime)s - [%(levelname)s] - %(message)s')
sns.set_theme(style="ticks")
plt.rcParams.update({'font.size': 9, 'axes.labelsize': 11, 'axes.titlesize': 12, 'figure.titlesize': 14})

def load_and_merge_data(f_heart="/content/heart.csv", f_o2="/content/o2Saturation.csv"):
    if not (os.path.exists(f_heart) and os.path.exists(f_o2)):
        raise FileNotFoundError("Fichiers manquants dans /content/")
    df = pd.read_csv(f_heart)
    df_o2 = pd.read_csv(f_o2)
    n_samples = len(df)
    df['Saturation_O2'] = df_o2.iloc[:n_samples, 0].values if len(df_o2) >= n_samples else np.random.normal(97.2, 1.2, n_samples)
    col_map = {
        'age': 'Age', 'sex': 'Sexe', 'cp': 'Type_Douleur', 'trtbps': 'Tension_Repos',
        'chol': 'Cholesterol', 'fbs': 'Glycemie_Ajeun', 'restecg': 'ECG_Repos',
        'thalachh': 'FC_Max', 'exng': 'Angine_Exercice', 'oldpeak': 'Depression_ST',
        'slp': 'Pente_ST', 'caa': 'Vaisseaux_Fluro', 'thall': 'Resultat_Thal', 'output': 'Target'
    }
    return df.rename(columns=col_map)

def clean_and_eng(df):
    res_df = df.copy()
    limit = res_df['Cholesterol'].mean() + (3 * res_df['Cholesterol'].std())
    bad_rows = res_df['Cholesterol'] > limit
    if bad_rows.any():
        res_df.loc[bad_rows, 'Cholesterol'] = res_df['Cholesterol'].median()
    res_df['Classe_Age'] = pd.cut(res_df['Age'], bins=[0, 45, 62, 110], labels=['<45 ans', '45-62 ans', '>62 ans'])
    res_df['Index_Charge_Physio'] = ((res_df['Tension_Repos'] * 0.35) + (res_df['Depression_ST'] * 12.0)).round(2)
    cat_cols = ['Sexe', 'Type_Douleur', 'Glycemie_Ajeun', 'ECG_Repos', 'Angine_Exercice', 'Target']
    res_df[cat_cols] = res_df[cat_cols].astype('category')
    return res_df

def run_eda_plots(df):
    plt.figure(figsize=(9, 6.5))
    num_cols = ['Age', 'Tension_Repos', 'Cholesterol', 'FC_Max', 'Depression_ST', 'Saturation_O2']
    calc_df = df[num_cols].copy()
    calc_df['Target'] = df['Target'].astype(int)
    c_matrix = calc_df.corr(method='pearson')
    mask = np.triu(np.ones_like(c_matrix, dtype=bool))
    sns.heatmap(c_matrix, mask=mask, annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.4, cbar_kws={"shrink": .8})
    plt.title('Inter-corrélations des paramètres cliniques', weight='bold', pad=12)
    plt.tight_layout()
    plt.show()

    fig, axes = plt.subplots(2, 2, figsize=(14, 9.5))
    fig.suptitle('Profils Physiologiques & Indicateurs de Risque (Kaggle Source)', weight='bold', y=0.97)
    sns.histplot(data=df, x='FC_Max', hue='Target', ax=axes[0, 0], palette='Set1', multiple='stack', alpha=0.7, kde=True)
    axes[0, 0].set(title='Distribution de la FC Max globale', xlabel='FC Max (BPM)', ylabel='Effectif')
    sns.scatterplot(data=df, x='Age', y='Saturation_O2', hue='Target', style='Angine_Exercice', ax=axes[0, 1], palette='deep', s=40)
    axes[0, 1].set(title='Dispersion : Saturation O₂ vs Âge du patient', xlabel='Âge (années)', ylabel='SpO₂ (%)')
    sns.boxplot(data=df, x='Classe_Age', y='Index_Charge_Physio', hue='Target', ax=axes[1, 0], palette='muted', width=0.45)
    axes[1, 0].set(title='Index de Charge Cardio par Tranche Clinique', xlabel='Groupes d\'âge', ylabel='Index (Tension + ST)')
    tmp_df = df.copy()
    tmp_df['Target'] = tmp_df['Target'].astype(int)
    sns.kdeplot(data=tmp_df, x='Tension_Repos', y='Cholesterol', hue='Target', ax=axes[1, 1], palette='inferno', alpha=0.5, levels=5)
    axes[1, 1].set(title='Densité de probabilité : Tension vs Cholestérol', xlabel='Tension au repos (mmHg)', ylabel='Cholestérol (mg/dl)')
    plt.tight_layout(rect=[0, 0.03, 1, 0.94])
    plt.show()

    plt.figure(figsize=(9, 4.5))
    sns.barplot(data=df, x='Type_Douleur', y=df['Target'].astype(int), hue='Glycemie_Ajeun', errorbar=None, palette='viridis', edgecolor='black', alpha=0.8)
    plt.title('Taux de positivité par type de douleur thoracique et glycémie', weight='bold', pad=12)
    plt.xlabel('Type de douleur (0 à 3)')
    plt.ylabel('Taux de risque moyen')
    plt.legend(title='Glycémie > 120mg/dl (1=Oui)')
    sns.despine(trim=True)
    plt.tight_layout()
    plt.show()

if __name__ == "__main__":
    raw_data = load_and_merge_data()
    clean_data = clean_and_eng(raw_data)
    print(f"• Échantillon analysé : {clean_data.shape[0]} patients.")
    print(f"• Taux global de positivité (Target) : {clean_data['Target'].astype(int).mean()*100:.1f}%")
    print("\n• Distribution du risque (%) selon le Type de Douleur Thoracique :")
    cross_pain = pd.crosstab(clean_data['Type_Douleur'], clean_data['Target'], normalize='index') * 100
    print(cross_pain.round(1).to_string())
    print("\n• Profils médians par groupe (Sains [0] vs Risque [1]) :")
    v_cols = ['Age', 'Tension_Repos', 'Cholesterol', 'FC_Max', 'Saturation_O2']
    print(clean_data.groupby('Target', observed=False)[v_cols].median().round(1))
    run_eda_plots(clean_data)
```
